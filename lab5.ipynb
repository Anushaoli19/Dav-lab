```python
#DETECTION USING IQR OUTLIER METHOD

import pandas as pd

# 1. Read the CSV file
df = pd.read_csv("students.csv")

# 2. Select numerical columns
numeric_cols = df.select_dtypes(include="number").columns

# 3. Detect outliers using IQR
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    # Identify outliers
    outliers_iqr = df[col][
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(f"\nColumn: {col}")
    print(f"IQR bounds: [{lower_limit:.2f}, {upper_limit:.2f}]")
    print("Outliers found (IQR):", len(outliers_iqr))
    print("Outlier values:", outliers_iqr.tolist())
```

```python
#DETECTION USING Z-SCORE OUTLIER METHOD

import pandas as pd

# 1. Read the CSV file
df = pd.read_csv("students.csv")

# 2. Select numerical columns
numeric_cols = df.select_dtypes(include="number").columns

# 3. Detect outliers using Z-score
for col in numeric_cols:
    mean = df[col].mean()
    std = df[col].std()

    if pd.isna(std) or std == 0:
        print(f"{col}: Z-score cannot be calculated")
        continue

    z_scores = (df[col] - mean) / std

    # Identify outliers
    outliers_z = df[col][z_scores.abs() > 3]

    print(f"\nColumn: {col}")
    print(f"Outliers found (Z-score): {len(outliers_z)}")
    print("Outlier values:", outliers_z.tolist())
```

```python
#TREATMENT USING IQR OUTLIER METHOD

import pandas as pd

# 1. Read the CSV file
df = pd.read_csv("students.csv")

# 2. Select numerical columns
numeric_cols = df.select_dtypes(include="number").columns

# 3. Detect outliers before capping
print("=== BEFORE CAPPING (IQR) ===")

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[col][(df[col] < lower) | (df[col] > upper)]

    print(f"\nColumn: {col}")
    print(f"IQR bounds: [{lower:.2f}, {upper:.2f}]")
    print("Outliers found:", len(outliers))

# 4. Apply capping
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower=lower, upper=upper)

# 5. Detect outliers after capping
print("\n=== AFTER CAPPING (IQR) ===")

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[col][(df[col] < lower) | (df[col] > upper)]

    print(f"{col}: {len(outliers)} outliers remaining")

# 6. Save treated data
df.to_csv("students_iqr_treated.csv", index=False)

print("\nIQR treatment completed!")
print(df.head())
```

```python
#TREATMENT USING Z-SCORE OUTLIER METHOD

import pandas as pd

# 1. Read the CSV file
df = pd.read_csv("students.csv")

# 2. Select numerical columns
numeric_cols = df.select_dtypes(include="number").columns

# 3. Detect outliers 
print("=== BEFORE CAPPING (Z-score) ===")

for col in numeric_cols:
    mean = df[col].mean()
    std = df[col].std()

    if pd.isna(std) or std == 0:
        print(f"{col}: Z-score cannot be calculated")
        continue

    z_scores = (df[col] - mean) / std
    outliers = df[col][z_scores.abs() > 3]

    print(f"\nColumn: {col}")
    print(f"Mean: {mean:.2f}, Standard deviation: {std:.2f}")
    print("Outliers found:", len(outliers))

# 4. Apply capping
for col in numeric_cols:
    mean = df[col].mean()
    std = df[col].std()

    if pd.isna(std) or std == 0:
        continue

    lower = mean - 3 * std
    upper = mean + 3 * std

    df[col] = df[col].clip(lower=lower, upper=upper)

# 5. Detect outliers after capping
print("\n=== AFTER CAPPING (Z-score) ===")

for col in numeric_cols:
    mean = df[col].mean()
    std = df[col].std()

    if pd.isna(std) or std == 0:
        print(f"{col}: Z-score cannot be calculated")
        continue

    z_scores = (df[col] - mean) / std
    outliers = df[col][z_scores.abs() > 3]

    print(f"{col}: {len(outliers)} outliers remaining")

# 6. Save treated data
df.to_csv("students_zscore_treated.csv", index=False)

print("\nZ-score treatment completed!")
print(df.head())
```

```python
#BOTH IQR AND Z-SCORE METHOD 
import pandas as pd

# 1. Read the CSV file
df = pd.read_csv("students.csv")

# Exclude StudentID because it is an identifier, not a measurement
numeric_cols = df.select_dtypes(include="number").columns.drop(
    "StudentID", errors="ignore"
)

# Keep a copy of the original data for both methods
df_iqr = df.copy()
df_zscore = df.copy()


# ==========================================
# PART 1: IQR DETECTION AND TREATMENT
# ==========================================

print("========== IQR METHOD: BEFORE CAPPING ==========")

for col in numeric_cols:
    Q1 = df_iqr[col].quantile(0.25)
    Q3 = df_iqr[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers_iqr = df_iqr[col][
        (df_iqr[col] < lower) | (df_iqr[col] > upper)
    ]

    print(f"\nColumn: {col}")
    print(f"IQR bounds: [{lower:.2f}, {upper:.2f}]")
    print("Outliers found:", len(outliers_iqr))
    print("Outlier values:", outliers_iqr.tolist())

# Apply IQR capping
for col in numeric_cols:
    Q1 = df_iqr[col].quantile(0.25)
    Q3 = df_iqr[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_iqr[col] = df_iqr[col].clip(lower=lower, upper=upper)

print("\n========== IQR METHOD: AFTER CAPPING ==========")

for col in numeric_cols:
    Q1 = df_iqr[col].quantile(0.25)
    Q3 = df_iqr[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers_iqr = df_iqr[col][
        (df_iqr[col] < lower) | (df_iqr[col] > upper)
    ]

    print(f"{col}: {len(outliers_iqr)} outliers remaining")

df_iqr.to_csv("students_iqr_treated.csv", index=False)


# ==========================================
# PART 2: Z-SCORE DETECTION AND TREATMENT
# ==========================================

print("\n========== Z-SCORE: BEFORE CAPPING ==========")

for col in numeric_cols:
    mean = df_zscore[col].mean()
    std = df_zscore[col].std()

    if pd.isna(std) or std == 0:
        print(f"\n{col}: Z-score cannot be calculated")
        continue

    z_scores = (df_zscore[col] - mean) / std
    outliers_z = df_zscore[col][z_scores.abs() > 3]

    print(f"\nColumn: {col}")
    print(f"Mean: {mean:.2f}, Standard deviation: {std:.2f}")
    print("Outliers found (Z-score):", len(outliers_z))
    print("Outlier values:", outliers_z.tolist())

# Apply Z-score capping
for col in numeric_cols:
    mean = df_zscore[col].mean()
    std = df_zscore[col].std()

    if pd.isna(std) or std == 0:
        continue

    lower = mean - 3 * std
    upper = mean + 3 * std

    df_zscore[col] = df_zscore[col].clip(
        lower=lower, upper=upper
    )

print("\n========== Z-SCORE: AFTER CAPPING ==========")

for col in numeric_cols:
    mean = df_zscore[col].mean()
    std = df_zscore[col].std()

    if pd.isna(std) or std == 0:
        print(f"{col}: Z-score cannot be calculated")
        continue

    z_scores = (df_zscore[col] - mean) / std
    outliers_z = df_zscore[col][z_scores.abs() > 3]

    print(f"{col}: {len(outliers_z)} outliers remaining")

df_zscore.to_csv("students_zscore_treated.csv", index=False)


# ==========================================
# FINAL OUTPUT
# ==========================================

print("\nIQR-treated data shape:", df_iqr.shape)
print("Z-score-treated data shape:", df_zscore.shape)

print("\nIQR-treated data:")
print(df_iqr.head())

print("\nZ-score-treated data:")
print(df_zscore.head())

print("\nBoth methods completed successfully!")
```
